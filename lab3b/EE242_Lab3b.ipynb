```python
# Assignment 4 - Comparing frequency content of a signal

# Part A - Reading, plotting and playing the bluenose3.wav file
file_path = 'bluenose3.wav'
fs, signal_data = read(file_path)
signal = signal_data.astype('float64')
print(f"Original Sample Rate: {fs} Hz")
print("Playing bluenose3 audio:")
ipd.display(ipd.Audio(signal, rate=fs))
time_axis = np.arange(len(signal)) / fs
plt.figure(figsize=(12, 4))
plt.plot(time_axis, signal)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title('Full Waveform of bluenose3')
plt.grid()
plt.show()

def signal_processing(t_start, t_end, signal, fs, title):
    # Extract the segment
    start_idx = int(t_start * fs)
    end_idx = int(t_end * fs)
    segment = signal[start_idx:end_idx]
    n = len(segment)  # Size of FFT
    
    # Compute FFT and frequency axis
    freq = np.fft.fftfreq(n, 1/fs)
    fft_result = np.abs(fft(segment))
    
    # Plot time and frequency domains
    plt.figure(figsize=(12, 8))
    
    # Time domain plot
    plt.subplot(2, 1, 1)
    t = np.linspace(t_start, t_end, len(segment))
    plt.plot(t, segment)
    plt.xlabel('Time (s)')
    plt.ylabel('Amplitude')
    plt.title(f'{title} - Time Domain')
    plt.grid()
    
    # Frequency domain plot
    plt.subplot(2, 1, 2)
    pos_freq = freq >= 0
    plt.plot(freq[pos_freq], fft_result[pos_freq])
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.title(f'{title} - Frequency Domain')
    plt.grid()
    
    return n
# Part B - Extracting ooo part of the signal
# Extract the ooo part of the signal from 0.75 seconds to 0.78 seconds
# Plot the signal in time and frequency domain
ooo_part = signal_processing(0.75, 0.78, signal, fs, 'ooo Signal')

# Part C - Extracting sss part of the signal
# Extract the sss part of the signal from 2.565 seconds to 2.595 seconds
# Plot the signal in time and frequency domain
sss_part = signal_processing(2.565, 2.595, signal, fs, 'sss Signal')

plt.show()
# Part D - Justify the difference and choice of nfft used.
```
